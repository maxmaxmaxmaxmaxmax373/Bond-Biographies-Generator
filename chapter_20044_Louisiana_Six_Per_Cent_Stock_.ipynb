## Approximate Yield to Maturity

Using the bond's coupon rate (6.0%) and remaining years to maturity at each observation, the chart below shows an approximate YTM trajectory. Higher YTM indicates lower market confidence or higher required returns; lower YTM indicates the opposite.

```python
# Approximate Yield to Maturity
p = BondPrice_T.loc[(20044, 'Average')]
p = pd.to_numeric(p, errors='coerce').dropna()

final_redemption = pd.Timestamp('1823-10-23')
coupon_rate = 6.0

ytm_values = []
ytm_dates = []
for date, price in p.items():
    years_left = (final_redemption - date).days / 365.25
    if years_left > 0 and price > 0:
        ytm = (coupon_rate + (100 - price) / years_left) / ((100 + price) / 2) * 100
        ytm_values.append(ytm)
        ytm_dates.append(date)

if ytm_values:
    fig, ax = plt.subplots(figsize=(12, 5))
    ax.plot(ytm_dates, ytm_values, color='#9b59b6', marker='.', markersize=3, linewidth=1.5)
    ax.axhline(6.0, color='black', alpha=0.2, linestyle='--', label=f'Coupon Rate ({coupon_rate}%)')
    ax.set_title('Louisiana Six Per Cent. Stock  — Approximate Yield to Maturity')
    ax.set_xlabel('Date')
    ax.set_ylabel('YTM (%)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
else:
    print('Could not compute YTM')
```
